```python
import pandas as pd
df = pd.read_csv("Q07_baggage_handling.csv")
print(df.head())
```

```python
df["transfer_count"] = df["transfer_count"].fillna(0)

df = df.drop_duplicates("baggage_tag")
```

```python
df["mishandled"] = df["mishandled"].map({"Yes": 1, "No": 0})
print(df.groupby("route_type")["mishandled"].mean() * 100)
print(df.groupby("transfer_count")["mishandled"].mean() * 100)
```

```python
df["complexity_score"] = (
    df["transfer_count"] * df["route_distance_km"]
)
```

```python
df["distance_group"] = pd.cut(
    df["route_distance_km"],
    bins=[0, 2000, 5000, float("inf")],
    labels=["Short", "Medium", "Long"]
)
```
